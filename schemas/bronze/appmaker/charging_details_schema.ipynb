```python
from pyspark.sql.types import (
    StructType,
    StructField,
    StringType,
    TimestampType,
    DecimalType,)
```

```python
charging_details_schema = StructType([
    StructField("datetime_ts",       TimestampType(),    True),
    StructField("transaction_id",    StringType(),       True),
    StructField("msisdn",            StringType(),       True),  # always string, never numeric
    StructField("service_name",      StringType(),       True),
    StructField("telco",             StringType(),       True),
    StructField("sub_service_name",  StringType(),       True),
    StructField("service_provider",  StringType(),       True),
    StructField("transaction_type",  StringType(),       True),
    StructField("platform_provider", StringType(),       True),
    StructField("app_id",            StringType(),       True),
    StructField("charged_amount",    DecimalType(10, 2), True),
    StructField("status_of_attempt", StringType(),       True),
    StructField("service_id",        StringType(),       True),
])

```
